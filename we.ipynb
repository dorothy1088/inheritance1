```python
class Vehicle:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def drive(self, km):
        print(f'{self.brand} {self.model}이(가) {km} km 주행했습니다.')

sunata = Vehicle('흔다이','소나타',2026)
sunata.drive(100)
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def sleep(self, time):
        print(f'{self.name}가 {time}만큼 잠을 잡니다.')

class Dog(Animal):
    def bark(self, time):
        print(f'{time}번 짖습니다. 멍!')

class Cat(Animal):
    def play(self, time):
        print(f'{time}시간만큼 혼자 놀았습니다.')
```

```python
dog = Dog('멍멍이')
cat = Cat('나비')

def care(animal: Animal, time: int) -> None:
    animal.sleep(time)

care(dog, 10)
care(cat, 10)
```
